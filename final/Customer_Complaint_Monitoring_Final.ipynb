## 5. Evaluation Metrics

To evaluate the candidate models and baseline models, we apply metrics as follows:

1. F1-score - Macro, Positive, Negative, Neutral
2. Accuracy
3. Precision - Positive, Negative, Neutral
4. Recall - Positive, Negative, Neutral
5. Confusion Matrix
6. Inference speed

### Purpose and use of each evaluation metrics

1. F1-score: This metric is calculated by combining precision and recall. Thus, to achieve high F1-score, both precision and recall should be evenly high.
We use Macro, Positive, Negative, and Neutral to inspect our model's performance in different aspects. Macro F1 for overall, positive F1 for focused view on positive predictions, negative F1 for focused view on negatitve predictions, neutral F1 for focused view on neutral predictions.

2. Accuracy: Accuracy is used to inspect how much the model predicted correctly among the whole data.

3. Precision: Precision is calculated for positive, negative, and neutral.
Precision refers to ratio of actual positives among all predicted positives.
This applies the same to negative and neutral.

4. Recall: Recall is also calculated for all positive, negative, and neutral.
This indicates how many cases were predicted positive among actual positive data. This applies the same to negative and neutral.

5. Confusion Matrix: This table shows which predictions were correct
and which classes were confused. Rows show actual labels, and columns
show predicted labels. The three-class models use a 3 × 3 matrix.
GPT-2 uses a 2 × 2 matrix because it evaluates only negative and
positive tweets.

6. Prediction Speed: Number of posts and comments on the internet is rapidly rising. Fast prediction speed is important to follow the trend and ultimately to reflect user's needs and gain useful feedbacks from customers.
By improving the prediction speed, models could be more suitable to follow rapidly changing environment.
By evaluating the speed, we expect to inspect and compare which model is more suitable in the perspective of swiftness.
To inspect in various perspectives, we used Total prediction time (How long it took to handle for the whole data), Available examples per second (How many samples can be handled in a second), and Average time cost per example (How long it took to handle for one example - millisecond).

```python
import seaborn as sns

from sklearn.metrics import accuracy_score, precision_recall_fscore_support, f1_score, confusion_matrix

label_ids = [0, 1, 2]
label_names = ["Negative", "Neutral", "Positive"]

def evaluate_model(model_name, y_true, y_pred, prediction_time=np.nan, supported_labels=None):
  if supported_labels is None:
    supported_labels = label_ids

  y_true = np.asarray(y_true)
  y_pred = np.asarray(y_pred)

  if len(y_true) != len(y_pred):
    raise ValueError(f"{model_name}: y_true and y_pred lengths differ ({len(y_true)} != {len(y_pred)}).")

  precision, recall, f1, support = precision_recall_fscore_support(
    y_true, y_pred, labels=supported_labels, zero_division=0
  )

  precision_by_label = dict(zip(supported_labels, precision))
  recall_by_label = dict(zip(supported_labels, recall))
  f1_by_label = dict(zip(supported_labels, f1))

  valid_time = prediction_time is not None and np.isfinite(prediction_time) and prediction_time > 0

  result = {
    "Model": model_name,
    "Evaluation Scope": "Negative + Neutral + Positive" if 1 in supported_labels else "Negative + Positive only",
    "Number of Examples": len(y_true),
    "Accuracy": accuracy_score(y_true, y_pred),
    "Macro F1": f1_score(y_true, y_pred, labels=supported_labels, average="macro", zero_division=0),
    "Negative F1": f1_by_label.get(0, np.nan),
    "Neutral F1": f1_by_label.get(1, np.nan),
    "Positive F1": f1_by_label.get(2, np.nan),
    "Negative Precision": precision_by_label.get(0, np.nan),
    "Neutral Precision": precision_by_label.get(1, np.nan),
    "Positive Precision": precision_by_label.get(2, np.nan),
    "Negative Recall": recall_by_label.get(0, np.nan),
    "Neutral Recall": recall_by_label.get(1, np.nan),
    "Positive Recall": recall_by_label.get(2, np.nan),
    "Total prediction time (seconds)": prediction_time if valid_time else np.nan,
    "Available examples per second": len(y_true) / prediction_time if valid_time else np.nan,
    "Avg time cost per example (milliseconds)": prediction_time * 1000 / len(y_true) if valid_time else np.nan
  }

  cm = confusion_matrix(y_true, y_pred, labels=label_ids)
  return result, cm

true_labels = test_df["label"].to_numpy()
binary_mask = true_labels != 1
true_labels_gpt2 = true_labels[binary_mask]
predictions_gpt2_common = np.where(np.asarray(predictions_gpt2) == 0, 0, 2)

prediction_times = {
  "RoBERTa": globals().get("prediction_time_twitter", np.nan),
  "GPT-2": globals().get("prediction_time_gpt2", np.nan),
  "BERTweet": globals().get("prediction_time_bertweet", np.nan),
  "DistilBERT": globals().get("prediction_time_distilbert", np.nan)
}

roberta_result, roberta_cm = evaluate_model(
  "RoBERTa", true_labels, predictions_twitter, prediction_times["RoBERTa"]
)

gpt2_result, gpt2_cm = evaluate_model(
  "GPT-2", true_labels_gpt2, predictions_gpt2_common, prediction_times["GPT-2"], supported_labels=[0, 2]
)

bertweet_result, bertweet_cm = evaluate_model(
  "BERTweet", true_labels, predictions_bertweet, prediction_times["BERTweet"]
)

distilbert_result, distilbert_cm = evaluate_model(
  "DistilBERT", true_labels, predictions_distilbert, prediction_times["DistilBERT"]
)

evaluation_results = pd.DataFrame([
  roberta_result, gpt2_result, bertweet_result, distilbert_result
])

numeric_columns = evaluation_results.select_dtypes(include=[np.number]).columns
evaluation_results[numeric_columns] = evaluation_results[numeric_columns].round(4)
display(evaluation_results)

confusion_matrices = {
  "RoBERTa": (roberta_cm, ["Negative", "Neutral", "Positive"]),
  "GPT-2 (Negative/Positive only)": (gpt2_cm[np.ix_([0, 2], [0, 2])], ["Negative", "Positive"]),
  "BERTweet": (bertweet_cm, ["Negative", "Neutral", "Positive"]),
  "DistilBERT": (distilbert_cm, ["Negative", "Neutral", "Positive"])
}

fig, axes = plt.subplots(2, 2, figsize=(14, 11))

for ax, (model_name, (cm, cm_labels)) in zip(axes.flatten(), confusion_matrices.items()):
  sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=cm_labels,
    yticklabels=cm_labels,
    ax=ax
  )

  ax.set_title(model_name)
  ax.set_xlabel("Predicted")
  ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

## 8. Metadata-Based Performance Analysis

We compare the production model, Twitter RoBERTa, with BERTweet
and DistilBERT across tweet-length groups and tweets with or without
mentions, URLs and hashtags.

We report group sizes, negative precision and negative recall.
GPT-2 is excluded from this direct comparison because its evaluation
covers only negative and positive tweets.


```python
import numpy as np
import pandas as pd

# Preserve the original evaluation data while adding analysis groups.
analysis_df = eval_df.copy()

# Use fixed length ranges for descriptive analysis.
# These ranges do not change model inputs or predictions.
analysis_df["word_group"] = pd.cut(
    analysis_df["word_count"],
    bins=[-np.inf, 10, 20, np.inf],
    labels=[
        "Short: 0–10 words",
        "Medium: 11–20 words",
        "Long: 21+ words"
    ]
)

analysis_df["character_group"] = pd.cut(
    analysis_df["character_count"],
    bins=[-np.inf, 80, 120, np.inf],
    labels=[
        "0–80 characters",
        "81–120 characters",
        "121+ characters"
    ]
)

# Compare models evaluated on the same full three-class test set.
# GPT-2 is excluded here because it was evaluated on a binary subset.
models = {
    "RoBERTa": "roberta_pred",
    "BERTweet": "bertweet_pred",
    "DistilBERT": "distilbert_pred"
}

slice_columns = [
    "word_group",
    "character_group",
    "has_mention",
    "has_url",
    "has_hashtag"
]

rows = []

# Evaluate each model within each metadata group.
# observed=True excludes length categories containing no examples.
for feature in slice_columns:
    for group_name, group in analysis_df.groupby(
        feature, observed=True
    ):
        actual_negative = group["label"].eq(0)

        for model_name, prediction_column in models.items():
            assert group[prediction_column].notna().all(), (
                f"Missing predictions for {model_name}"
            )

            predicted_negative = group[prediction_column].eq(0)

            # Treat Negative as the target class and all other labels
            # as non-negative when calculating precision and recall.
            true_positives = int(
                (actual_negative & predicted_negative).sum()
            )
            actual_count = int(actual_negative.sum())
            predicted_count = int(predicted_negative.sum())

            # Undefined metrics remain NaN rather than being reported
            # as zero. Counts help identify groups with little evidence.
            negative_precision = (
                true_positives / predicted_count
                if predicted_count else np.nan
            )
            negative_recall = (
                true_positives / actual_count
                if actual_count else np.nan
            )

            rows.append({
                "Metadata": feature,
                "Group": str(group_name),
                "Model": model_name,
                "Tweets": len(group),
                "Actual negatives": actual_count,
                "Predicted negatives": predicted_count,
                "Negative precision": negative_precision,
                "Negative recall": negative_recall
            })

slice_results = pd.DataFrame(rows)

# Format metrics for display while keeping numeric values unchanged
# for later charts, exports and W&B logging.
display(
    slice_results.style.format(
        {
            "Negative precision": "{:.1%}",
            "Negative recall": "{:.1%}"
        },
        na_rep="N/A"
    )
)
```

### 8.1 Findings from Metadata-Based Analysis

We compared the models across tweet lengths, mentions, URLs and
hashtags to see where their performance changed.

#### Tweet length

All three models had lower negative precision and recall on short
tweets than on longer tweets. For tweets with 10 words or fewer,
BERTweet performed slightly better than RoBERTa. Its precision was
61.2%, compared with 58.1% for RoBERTa, while recall was similar
at 73.7% and 73.3%.

For medium and long tweets, BERTweet had higher precision but lower
recall than RoBERTa. Grouping tweets by character count showed a
similar pattern.

However, the groups had different proportions of negative tweets:
18.2% of short tweets were negative, compared with 49.4% of long
tweets. This difference can affect precision, so we cannot conclude
that length alone explains the results.

#### Mentions

All three models had higher negative precision on tweets containing
mentions. BERTweet was more precise than RoBERTa both with and without
mentions, but RoBERTa caught a slightly larger proportion of the
actual negative tweets in both groups.

#### URLs

RoBERTa and BERTweet performed worse on tweets containing URLs than
on tweets without them. Both had 72.3% negative recall in the URL
group. Their precision was 56.1% and 55.0%, respectively.

DistilBERT caught more negative tweets in this group, with 77.1%
recall, but its precision was only 41.0%. This means many of its
negative predictions were false alerts.

The URL group contained only 377 tweets, including 83 negatives,
so we should be cautious when drawing conclusions from it.

#### Hashtags

For tweets containing hashtags, BERTweet had higher negative
precision than RoBERTa: 62.6% compared with 59.7%. However, its
recall was lower: 79.5% compared with 83.2%.

This shows the trade-off between the two models. BERTweet made
fewer false alerts in this group, while RoBERTa missed fewer
negative tweets.

#### Overall findings

BERTweet usually improved negative precision, but it often reduced
recall compared with RoBERTa. DistilBERT generally had lower precision
than both models. None reached 80% negative precision in any of the
groups we examined.

These results help us identify where each model struggles, but they
do not prove that mentions, URLs, hashtags or tweet length cause
the errors. The groups overlap and contain different types of tweets.

## 9. Model Selection and Deployment Recommendation

### 9.1 Comparing the Models

Our goal was to find out whether BERTweet or DistilBERT could replace
Twitter RoBERTa, the current production model. We compared all three
models on the same 12,284 test tweets.

| Model | Accuracy | Macro F1 | Negative precision | Negative recall |
|---|---:|---:|---:|---:|
| Twitter RoBERTa | 72.18% | 0.7240 | 68.95% | 80.66% |
| BERTweet | 72.07% | 0.7207 | 70.76% | 79.38% |
| DistilBERT | 43.59% | 0.3935 | 53.16% | 78.42% |

RoBERTa had the highest macro F1 and caught the largest proportion of
negative tweets. BERTweet performed similarly overall and was more
precise when predicting negative sentiment. However, it missed more
negative tweets than RoBERTa.

DistilBERT had lower scores on all four measures, so our results do
not support choosing it as a replacement.

We kept GPT-2 as a separate comparison because it was tested only on
negative and positive tweets. Its results are not directly comparable
with the three-class results above.

### 9.2 Our Recommendation

We recommend keeping Twitter RoBERTa for now.

BERTweet reduced false negative-sentiment alerts overall, but this
improvement came with a reduction in negative recall. Since our
project aims to catch more negative feedback while maintaining
precision, this trade-off does not meet our target.

The metadata analysis showed a similar pattern. For tweets containing
hashtags, BERTweet achieved higher precision than RoBERTa (62.6%
versus 59.7%), but lower recall (79.5% versus 83.2%). It was therefore
not consistently better across the groups we examined.

### 9.3 Did We Meet Our Target?

We did not achieve our target of improving negative recall while
keeping negative precision at or above 80%. Neither candidate improved
overall negative recall over RoBERTa, and all three models remained
below the precision target.

Keeping RoBERTa does not mean it is fully reliable. Its negative
precision was 68.95%, so some tweets it flags as negative are actually
neutral or positive according to the dataset. We would use these
predictions to help prioritize human review, rather than automatically
treat every negative prediction as a confirmed complaint.

Finally, our evaluation used general tweets rather than a dataset of
verified customer complaints. Before using the system for a business,
we would need to test it on real customer feedback and check whether
these findings still hold.
